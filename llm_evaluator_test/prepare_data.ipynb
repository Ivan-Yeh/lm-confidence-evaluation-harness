```python
import pandas as pd
```

```python
full_df = pd.read_csv("full_dataset.csv")[["uncertainty_expression", "llm_confidence_level", "annotater_confidence_score1", "annotater_confidence_score2", "annotater_confidence_score3", "annotater_confidence_score4", "annotater_confidence_score5"]]
filtered_df = pd.read_csv("linguistic_confidence_benchmark.csv")[["uncertainty_expression", "llm_confidence_level", "annotater_confidence_score1", "annotater_confidence_score2", "annotater_confidence_score3", "annotater_confidence_score4", "annotater_confidence_score5"]]
```

```python
# Keep only full_df rows with matching uncertainty_expression in filtered_df
df = full_df[full_df["uncertainty_expression"].isin(filtered_df["uncertainty_expression"])].copy()
df
```

```python
consolidated_df = df[["uncertainty_expression", "llm_confidence_level"]].copy()
consolidated_df["annotated_scores"] = df[["annotater_confidence_score1", "annotater_confidence_score2", "annotater_confidence_score3", "annotater_confidence_score4", "annotater_confidence_score5"]].values.tolist()
consolidated_df["annotated_scores"] = consolidated_df["annotated_scores"].apply(lambda x: [y / 100 for y in x if not pd.isna(y)])
```

```python
consolidated_df.to_csv("annotated_confidence_data.pkl", index=True)
```
