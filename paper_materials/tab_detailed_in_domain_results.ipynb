```python
prompt_type = "direct_qa"
results_dir = f"/hdd/ivny/{prompt_type}_in_domain_calibration"

# list all leaf nodes in results_dir
leaf_dirs = []
for root, dirs, files in os.walk(results_dir):
    if not dirs:  # if there are no subdirectories, it's a leaf node
        leaf_dirs.append(root)

all_records = []
for leaf_dir in leaf_dirs:
    _, _, _, _, dataset_name, _, model_name = leaf_dir.split("/")
    if model_name == "gpt-oss-120b" or model_name == "Qwen3-235B-A22B-Instruct-2507-tput":
            continue
    if os.path.exists(os.path.join(leaf_dir, "calibration_performance.csv")):
        csv_data = pd.read_csv(os.path.join(leaf_dir, "calibration_performance.csv"), index_col=0)
        csv_records = csv_data.transpose().to_dict("records")

        for record in csv_records:
            record["dataset"] = dataset_map.get(dataset_name, dataset_name)
            record["model"] = model_name_map.get(model_name, model_name)
            record["dataset_size"] = dataset_size.get(dataset_name, None)

        all_records.extend(csv_records)
    else:
        print(model_name, dataset_name, "missing calibration_performance.csv")
```

```python
def get_latest_subdir(path):
    subdirs = [os.path.join(path, d) for d in os.listdir(path) if os.path.isdir(os.path.join(path, d))]
    return max(subdirs, key=os.path.getmtime) if subdirs else None

results_dir = "/hdd/ivny/results" if os.path.exists("/hdd/ivny") else "/home/ivan/lm-confidence-evaluation-harness/ivny/results"

paths = {
    "mmlu":        os.path.join(results_dir, "mmlu",        f"{prompt_type}_unified_lc"),
    "squadv2":     os.path.join(results_dir, "squadv2",     f"{prompt_type}_unified_lc"),
    "truthful_qa": os.path.join(results_dir, "truthful_qa", f"{prompt_type}_unified_lc"),
}

all_records_hedged = []
for dataset_name, path in paths.items():
    if not os.path.exists(path):
        print(f"Missing: {path}")
        continue
    seen_models = {}  # model_name -> (latest_dir, latest_mtime)
    for org in os.listdir(path):
        org_path = os.path.join(path, org)
        if not os.path.isdir(org_path):
            continue
        for model_name in os.listdir(org_path):
            model_path = os.path.join(org_path, model_name)
            if not os.path.isdir(model_path):
                continue
            latest = get_latest_subdir(model_path)
            if latest is None:
                continue
            mtime = os.path.getmtime(latest)
            if model_name not in seen_models or mtime > seen_models[model_name][1]:
                seen_models[model_name] = (latest, mtime)

    for model_name, (latest_dir, _) in seen_models.items():
        if model_name == "gpt-oss-120b" or model_name.lower() == "qwen3-235B-A22B-Instruct-2507-tput":
            continue
        csv_path = os.path.join(latest_dir, "eval_metrics.csv")
        if not os.path.exists(csv_path):
            print(f"Missing eval_metrics.csv: {latest_dir}")
            continue
        try:
            row = pd.read_csv(csv_path).iloc[0]
            all_records_hedged.append({
                "Model":                  model_name_map.get(model_name, model_name),
                "Dataset":                dataset_map.get(dataset_name, dataset_name),
                "Estimator":             "Linguistic Confidence",
                "dataset_size":           dataset_size.get(dataset_name, None),
                "Hedged QA Generalised ECE":        row["generalised_ece"],
                "Hedged QA Faithfulness Divergence": row["faithfulness_divergence"],
            })
            all_records_hedged.append({
                "Model":                  model_name_map.get(model_name, model_name),
                "Dataset":                dataset_map.get(dataset_name, dataset_name),
                "Estimator":             "Token Probability",
                "dataset_size":           dataset_size.get(dataset_name, None),
                "Hedged QA Generalised ECE":        row["generalised_ece"],
                "Hedged QA Faithfulness Divergence": row["faithfulness_divergence"],
            })
            all_records_hedged.append({
                "Model":                  model_name_map.get(model_name, model_name),
                "Dataset":                dataset_map.get(dataset_name, dataset_name),
                "Estimator":             "Semantic Uncertainty",
                "dataset_size":           dataset_size.get(dataset_name, None),
                "Hedged QA Generalised ECE":        row["generalised_ece"],
                "Hedged QA Faithfulness Divergence": row["faithfulness_divergence"],
            })
        except Exception as e:
            print(f"Error processing {csv_path}: {e}")

hedged_qa_df = pd.DataFrame(all_records_hedged)
```

```python
from collections import defaultdict


ESTIMATOR_ORDER = ["Linguistic Confidence", "Semantic Uncertainty", "Token Probability"]

ESTIMATOR_LABELS = {
    "Linguistic Confidence": "Linguistic Conf.",
    "Semantic Uncertainty":  "Semantic Unc.",
    "Token Probability":     "Token Prob.",
}


def _fmt(value: float, decimals: int = 3) -> str:
    return f"{value:.{decimals}f}"


def _arrow(orig: float, calib: float, decimals: int = 3) -> str:
    return f"${_fmt(orig, decimals)} \\to {_fmt(calib, decimals)}$"


def generate_latex_table(
    data: list[dict],
    caption: str = "Calibration results across datasets, models, and estimators. Lower is better for all metrics.",
    label: str = "tab:results",
    decimals: int = 3,
    dataset_order: list[str] | None = None,
    model_order: list[str] | None = None,
) -> str:
    """
    Convert a list of result dicts into a LaTeX table string.

    Args:
        data:          List of result dicts (see module docstring for expected keys).
        caption:       Table caption.
        label:         LaTeX label for \\ref{}.
        decimals:      Decimal places for all numeric values.
        dataset_order: Optional explicit ordering of datasets.
                       Defaults to insertion order found in data.
        model_order:   Optional explicit ordering of models within each dataset.
                       Defaults to insertion order found in data.

    Returns:
        A self-contained LaTeX string (table* environment).
        Required packages: booktabs, multirow, makecell.
    """
    # ── 1. Index data: dataset -> model -> estimator -> row ──────────────────
    index: dict[str, dict[str, dict[str, dict]]] = defaultdict(lambda: defaultdict(dict))
    seen_datasets: list[str] = []
    seen_models: dict[str, list[str]] = defaultdict(list)

    for row in data:
        ds   = row["Dataset"]
        mdl  = row["Model"]
        est  = row["Estimator"]

        if ds not in seen_datasets:
            seen_datasets.append(ds)
        if mdl not in seen_models[ds]:
            seen_models[ds].append(mdl)

        index[ds][mdl][est] = row

    datasets = dataset_order if dataset_order is not None else seen_datasets
    
    def get_models(ds):
        if model_order is not None:
            return [m for m in model_order if m in index[ds]]
        return seen_models[ds]

    # ── 2. Build header ───────────────────────────────────────────────────────
    header = r"""\begin{table*}[ht]
\centering
\caption{""" + caption + r"""}
\label{""" + label + r"""}
\resizebox{\textwidth}{!}{%
\begin{tabular}{llllcccc}
\toprule
\multirow{3}{*}{Dataset}
  & \multirow{3}{*}{Model}
  & \multirow{3}{*}{Estimator}
  & \multicolumn{2}{c}{\textbf{In-Domain Calibration}}
  & \multicolumn{2}{c}{\textbf{Hedged QA}} \\
\cmidrule(lr){4-5} \cmidrule(lr){6-7}
  & & & \makecell{Faithfulness \\ Divergence \\ (Orig.\ $\to$ Calib.)}
       & \makecell{Generalised \\ ECE \\ (Orig.\ $\to$ Calib.)}
       & \makecell{Faithfulness \\ Divergence \\ (Orig.\ $\to$ Hedged)}
       & \makecell{Generalised \\ ECE \\ (Orig.\ $\to$ Hedged)} \\
\midrule"""

    # ── 3. Build body rows ────────────────────────────────────────────────────
    body_lines: list[str] = []

    for ds_idx, ds in enumerate(datasets):
        models = get_models(ds)
        n_ds_rows = len(models) * len(ESTIMATOR_ORDER)

        ds_cell = f"\\multirow{{{n_ds_rows}}}{{*}}{{{ds}}}"
        first_ds_row = True

        for mdl_idx, mdl in enumerate(models):
            estimators = ESTIMATOR_ORDER

            mdl_cell = f"\\multirow{{{len(estimators)}}}{{*}}{{{mdl}}}"
            first_mdl_row = True

            for est_idx, est in enumerate(estimators):
                row = index[ds][mdl].get(est)
                if row is None:
                    continue

                # In-domain calibration columns (vary per estimator)
                fd_cell  = _arrow(row["Original Faithfulness Divergence"],
                                  row["Calibrated Faithfulness Divergence"], decimals)
                ece_cell = _arrow(row["Original Generalised ECE"],
                                  row["Calibrated Generalised ECE"], decimals)

                # Hedged QA columns (same for all estimators of this model/dataset)
                if first_mdl_row:
                    hfd_cell = (f"\\multirow{{{len(estimators)}}}{{*}}{{"
                                + _arrow(row["Original Faithfulness Divergence"],
                                         row["Hedged QA Faithfulness Divergence"], decimals)
                                + "}")
                    hece_cell = (f"\\multirow{{{len(estimators)}}}{{*}}{{"
                                 + _arrow(row["Original Generalised ECE"],
                                          row["Hedged QA Generalised ECE"], decimals)
                                 + "}")
                else:
                    hfd_cell  = ""
                    hece_cell = ""

                est_label = ESTIMATOR_LABELS.get(est, est)

                # Assemble leading cells
                if first_ds_row and first_mdl_row:
                    leading = f" {ds_cell} & {mdl_cell}"
                elif first_mdl_row:
                    leading = f" & {mdl_cell}"
                else:
                    leading = f" &"

                line = (f"{leading} & {est_label}"
                        f" & {fd_cell} & {ece_cell}"
                        f" & {hfd_cell} & {hece_cell} \\\\")

                body_lines.append(line)
                first_ds_row   = False
                first_mdl_row  = False

            # Separator between models (not after the last model in a dataset)
            if mdl_idx < len(models) - 1:
                body_lines.append(r"\cmidrule(l){2-7}")

        # Separator between datasets (not after the last one)
        if ds_idx < len(datasets) - 1:
            body_lines.append(r"\midrule")

    # ── 4. Footer ─────────────────────────────────────────────────────────────
    footer = r"""\bottomrule
\end{tabular}%
}
\end{table*}"""

    return "\n".join([header, *body_lines, footer])


print(generate_latex_table(combined_df.to_dict("records"), caption="Sample table.", label="tab:sample"))
```
