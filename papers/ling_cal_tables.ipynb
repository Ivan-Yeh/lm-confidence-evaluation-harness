```python
# /hdd/ivny/results/mmlu/dist_linguistic_confidence/meta-llama/Llama-3.1-8B-Instruct/2026-01-12_23-56-50
# /hdd/ivny/results/mmlu/dist_linguistic_confidence/meta-llama/Meta-Llama-3-8B-Instruct/2026-01-12_22-07-53
# /hdd/ivny/results/mmlu/dist_linguistic_confidence/qwen/Qwen2.5-7B-Instruct/2026-01-13_03-31-49
# /hdd/ivny/results/mmlu/dist_linguistic_confidence/qwen/Qwen3-8B/2026-01-13_05-14-17
# /hdd/ivny/results/mmlu/dist_linguistic_confidence/mistralai/Mistral-7B-Instruct-v0.3/2026-01-13_07-39-23
# /hdd/ivny/results/mmlu/dist_linguistic_confidence/openai/gpt-oss-20b/2026-01-13_01-59-48
# /hdd/ivny/results/mmlu/dist_lnll/meta-llama/Llama-3.1-8B-Instruct/2026-01-12_03-43-10
# /hdd/ivny/results/mmlu/dist_lnll/meta-llama/Meta-Llama-3-8B-Instruct/2026-01-12_01-55-40
# /hdd/ivny/results/mmlu/dist_lnll/qwen/Qwen2.5-7B-Instruct/2026-01-12_07-23-22
# /hdd/ivny/results/mmlu/dist_lnll/qwen/Qwen3-8B/2026-01-12_08-44-05
# /hdd/ivny/results/mmlu/dist_lnll/mistralai/Mistral-7B-Instruct-v0.3/2026-01-12_20-11-42
# /hdd/ivny/results/mmlu/dist_lnll/openai/gpt-oss-20b/2026-01-12_05-33-45
# /hdd/ivny/results/trivia_qa/dist_linguistic_confidence/meta-llama/Llama-3.1-8B-Instruct/2026-01-13_20-24-13
# /hdd/ivny/results/trivia_qa/dist_linguistic_confidence/meta-llama/Meta-Llama-3-8B-Instruct/2026-01-13_19-01-11
# /hdd/ivny/results/trivia_qa/dist_linguistic_confidence/qwen/Qwen2.5-7B-Instruct/2026-01-13_23-09-32
# /hdd/ivny/results/trivia_qa/dist_linguistic_confidence/qwen/Qwen3-8B/2026-01-14_01-02-48
# /hdd/ivny/results/trivia_qa/dist_linguistic_confidence/mistralai/Mistral-7B-Instruct-v0.3/2026-01-14_03-24-46
# /hdd/ivny/results/trivia_qa/dist_linguistic_confidence/openai/gpt-oss-20b/2026-01-13_21-48-52
# /hdd/ivny/results/trivia_qa/dist_lnll/meta-llama/Llama-3.1-8B-Instruct/2026-01-12_06-21-48
# /hdd/ivny/results/trivia_qa/dist_lnll/meta-llama/Meta-Llama-3-8B-Instruct/2026-01-12_01-55-31
# /hdd/ivny/results/trivia_qa/dist_lnll/qwen/Qwen2.5-7B-Instruct/2026-01-12_17-52-11
# /hdd/ivny/results/trivia_qa/dist_lnll/qwen/Qwen3-8B/2026-01-13_01-09-42
# /hdd/ivny/results/trivia_qa/dist_lnll/mistralai/Mistral-7B-Instruct-v0.3/2026-01-14_08-49-59
# /hdd/ivny/results/trivia_qa/dist_lnll/openai/gpt-oss-20b/2026-01-12_12-22-09

def find_leaf_dirs(root):
    leaf_dirs = []
    for dirpath, dirnames, filenames in os.walk(root):
        if not dirnames:  # no subdirectories → leaf
            leaf_dirs.append(dirpath)
    return leaf_dirs

leaf_dirs = find_leaf_dirs("/hdd/ivny/results")
for d in leaf_dirs:
    if any(x in d for x in ["mmlu", "trivia_qa"]):
        print(d)
```

```python
# Read linguistic_calibration_metrics.csv from all leaf directories
all_metrics = []

for leaf_dir in leaf_dirs:
    if any(x in leaf_dir for x in ["mmlu", "trivia_qa"]):
        metrics_file = os.path.join(leaf_dir, "linguistic_calibration_metrics.csv")
        
        if os.path.exists(metrics_file):
            try:
                df_metrics = pd.read_csv(metrics_file)
                
                # Extract path components for context
                parts = leaf_dir.split("/")
                task = parts[4] if len(parts) > 4 else "unknown"
                method = parts[5] if len(parts) > 5 else "unknown"
                org = parts[6] if len(parts) > 6 else "unknown"
                model = parts[7] if len(parts) > 7 else "unknown"
                timestamp = parts[8] if len(parts) > 8 else "unknown"
                
                # Add metadata columns
                df_metrics["task"] = task
                df_metrics["method"] = method
                df_metrics["organization"] = org
                df_metrics["model"] = model
                df_metrics["timestamp"] = timestamp
                df_metrics["path"] = leaf_dir
                
                all_metrics.append(df_metrics)
                print(f"✓ Loaded {metrics_file}")
            except Exception as e:
                print(f"✗ Error loading {metrics_file}: {e}")
        else:
            print(f"✗ File not found: {metrics_file}")

# Aggregate into one big table
if all_metrics:
    aggregated_metrics = pd.concat(all_metrics, ignore_index=True)
    print(f"\nAggregated shape: {aggregated_metrics.shape}")
    print(f"Unique tasks: {aggregated_metrics['task'].nunique()}")
    print(f"Unique methods: {aggregated_metrics['method'].nunique()}")
    print(f"Unique models: {aggregated_metrics['model'].nunique()}")
    print("\nFirst few rows:")
    display(aggregated_metrics.head(10))
else:
    print("No metrics files found!")
```

```python
# Pivot aggregated metrics by task, model, and method
# Create a summary table grouped by dataset, model, and estimation method

pivot_data = []

for (task, model, method), group in aggregated_metrics.groupby(['task', 'model', 'method']):
    row = {
        'dataset': task,
        'model': model,
        'method': method,
    }
    
    # Get all metric values for this group
    for _, metric_row in group.iterrows():
        metric_name = metric_row['metric']
        value = metric_row['value']
        row[metric_name] = value
    
    pivot_data.append(row)

summary_table = pd.DataFrame(pivot_data)

# Sort by dataset, then model, then method
summary_table = summary_table.sort_values(['dataset', 'model', 'method']).reset_index(drop=True)

print(f"Summary table shape: {summary_table.shape}")
print(f"\nColumns: {summary_table.columns.tolist()}")
print("\nSummary by Dataset and Model:")
display(summary_table)
```
