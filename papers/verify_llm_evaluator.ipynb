```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
human_scores = pd.read_csv("../llm_evaluator_test/annotated_confidence_data.csv")
api_scores = pd.read_csv("../llm_evaluator_test/api_eval_results.csv")
local_scores = pd.read_csv("../llm_evaluator_test/local_eval_results.csv")
```

```python
def remove_outliers(values):
    arr = np.asarray(values, dtype=float)
    if len(arr) == 0:
        return np.array([np.nan])
    q1, q3 = np.percentile(arr, [25, 75])
    iqr = q3 - q1
    if iqr == 0:
        return arr
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return arr[(arr >= lower) & (arr <= upper)]
```

```python
human_scores["annotated_scores"] = human_scores["annotated_scores"].apply(lambda x: eval(x))
human_scores["annotated_scores"] = human_scores["annotated_scores"].apply(remove_outliers)
human_scores["annotated_scores"] = human_scores["annotated_scores"].apply(lambda x: np.mean(x))
api_scores["scores"] = api_scores["scores"].apply(lambda x: eval(x))
local_scores["scores"] = local_scores["scores"].apply(lambda x: eval(x))
```

```python
human_scores["llm_scores"] = local_scores["scores"]
human_scores["llm_scores"] = human_scores["llm_scores"].apply(remove_outliers)
human_scores.dropna(inplace=True)
human_scores
```

```python
human_scores["llm_scores"] = human_scores["llm_scores"].apply(np.mean)
```

```python
# Convert to long format for seaborn
df_long = human_scores.melt(
    id_vars=['llm_confidence_level'],
    value_vars=['annotated_scores', 'llm_scores'],
    var_name='source',
    value_name='score'
)
df_long['source'] = df_long['source'].map({
    'annotated_scores': 'Surveyers',
    'llm_scores': 'LLMs',
})

# Define the specific order for confidence levels
level_order = ['completely uncertain', 'lowest', 'low', 'moderate', 'high']

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8,4))
sns.boxplot(
    x='llm_confidence_level',
    y='score',
    hue='source',
    data=df_long,
    showfliers=False,
    palette=['skyblue', 'lightgreen'],
    order=level_order
)

# plt.title('Annotated vs LLM Scores by Confidence Level')
plt.xlabel('Benchmark Confidence Level')
plt.ylabel('Confidence Score')
plt.legend(title='Source')
plt.savefig('annotated_vs_llm_scores_boxplots.pdf', dpi=300, bbox_inches='tight')
```

```python
# Suppose human_scores already has columns: ['annotated_scores', 'llm_scores', 'llm_confidence_level']

# NeurIPS-style plotting defaults
import matplotlib as mpl

mpl.rcParams.update({
    "font.family": "serif",
    "font.serif": ["DejaVu Serif"],
    "axes.labelsize": 6,
    "axes.titlesize": 7,
    "xtick.labelsize": 4,
    "ytick.labelsize": 4,
    "legend.fontsize": 4,
    "axes.linewidth": 0.8,
    "grid.linewidth": 0.5,
    "grid.alpha": 0.3,
    "savefig.dpi": 300,
})

# Define order of confidence levels
level_order = ['completely uncertain', 'lowest', 'low', 'moderate', 'high']

sns.set_theme(context="paper", style="whitegrid")
fig, ax = plt.subplots(figsize=(6, 3))

# Boxplot for human/annotated scores
sns.boxplot(
    x='llm_confidence_level',
    y='annotated_scores',
    data=human_scores,
    order=level_order,
    color='#c7dcef',
    showfliers=False,
    linewidth=0.8,
    ax=ax,
    boxprops=dict(edgecolor="#2b2b2b"),
    whiskerprops=dict(color="#2b2b2b", linewidth=0.8),
    capprops=dict(color="#2b2b2b", linewidth=0.8),
    medianprops=dict(color="#2b2b2b", linewidth=0.9),
)

# Compute mean LLM score per confidence level
llm_means = human_scores.groupby('llm_confidence_level')['llm_scores'].mean().reindex(level_order)

# Overlay mean LLM scores as points
ax.scatter(
    x=range(len(llm_means)),
    y=llm_means.values,
    color="#1f1f1f",
    s=60,
    zorder=10,
    label='Mean LLM Score',
    marker='*',
)

# Do not display x-axis label
# ax.set_xlabel('Benchmark Confidence Level', fontsize=8)
ax.set_xlabel('')
ax.set_ylabel('Confidence Score', fontsize=8)
ax.set_ylim(0, 1)
ax.legend(frameon=True, fontsize=8)

# Split the longest label onto two lines
tick_labels = ["completely\nuncertain" if x == "completely uncertain" else x for x in level_order]
ax.set_xticklabels(tick_labels, fontsize=8)
ax.tick_params(axis='y', labelsize=8)

fig.tight_layout()
fig.savefig('annotated_vs_llm_scores_boxplots_1.pdf', bbox_inches='tight', dpi=300)
```
